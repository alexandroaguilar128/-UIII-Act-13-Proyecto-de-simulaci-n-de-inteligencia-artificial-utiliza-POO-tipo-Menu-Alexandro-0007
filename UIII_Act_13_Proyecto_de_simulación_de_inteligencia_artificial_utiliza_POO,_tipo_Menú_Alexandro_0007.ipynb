### Conceptos
#### Atributos:
* self.banco = Es donde se guarda el nombre del banco.
* self.titular = Es el nombre de la persona dueña de la cuenta de banco.
* self.n_cuenta = Es donde se guarda el numero de cuenta.
* self.saldo = Es la cantidad de dinero en la cuenta.
* self.moneda = Es donde se guarda la moneda que fue ingresada a la cuenta.

#### Funciones:
* __init__ = Permite el uso de los atributos en el codigo.
* Consultar_saldo = es donde se consulta la cantidad de dinero en la cuenta.
* depositar = Es el monto de dinero a ingresar a la cuenta de banco.
* retirar = Es el monto a retirar/sacar de dinero de la cuenta de banco.
* ver_datos = Es como se muestra la informacion de la cuenta, propietario, numero de cuenta y estado de cuenta.

```python
class cajero_automatico:
    # 1. Función Constructor: Inicializa los 5 atributos [6, 7]
    def __init__(self, banco, titular, n_cuenta, saldo, moneda):
        self.banco = banco           # Atributo 1
        self.titular = titular       # Atributo 2
        self.n_cuenta = n_cuenta     # Atributo 3
        self.saldo = saldo           # Atributo 4
        self.moneda = moneda         # Atributo 5

    # 2. Función para mostrar saldo [8]
    def consultar_saldo(self):
        print(f"\nSaldo actual: {self.saldo} {self.moneda}")

    # 3. Función para depositar dinero [9]
    def depositar(self, cantidad):
        self.saldo = self.saldo + cantidad
        print(f"Depósito exitoso. Nuevo saldo: {self.saldo}")

    # 4. Función para retirar dinero con validación [1, 9, 10]
    def retirar(self, cantidad):
        if cantidad <= self.saldo:
            self.saldo = self.saldo - cantidad
            print(f"Retiro realizado. Saldo restante: {self.saldo}")
        else:
            print("Error: Fondos insuficientes.")

    # 5. Función para ver datos de la cuenta [11, 12]
    def ver_datos(self):
        print("\n--- DETALLES DE LA CUENTA ---")
        print(f"Banco: {self.banco}")
        print(f"Titular: {self.titular}")
        print(f"Número: {self.n_cuenta}")
        print(f"Estado: Activa")
        print("-----------------------------")

# --- Lógica del Programa (Menú Interactivo) ---

# Instanciación del objeto [13, 14]
mi_cajero = cajero_automatico("Banco Central", "Juan Pérez", "987654", 1000.0, "MXN")

print(f"*** BIENVENIDO A {mi_cajero.banco} ***")

# Ciclo infinito para el menú [15, 16]
while True:
    print("\nMENÚ DE OPERACIONES:")
    print("1. Consultar saldo")
    print("2. Realizar depósito")
    print("3. Realizar retiro")
    print("4. Ver información de cuenta")
    print("5. Salir")
    opcion = input("Seleccione una opción (1-5): ") # [17, 18]

    # Estructura selectiva de casos [3, 19]
    if opcion == "1":
        mi_cajero.consultar_saldo()

    elif opcion == "2":
        monto = float(input("Monto a depositar: ")) # [20, 21]
        mi_cajero.depositar(monto)

    elif opcion == "3":
        monto = float(input("Monto a retirar: "))
        mi_cajero.retirar(monto)

    elif opcion == "4":
        mi_cajero.ver_datos()

    elif opcion == "5":
        print("Gracias por usar nuestros servicios. ¡Adiós!")
        break # Finaliza el ciclo [15]

print("Elaborado por Alexandro Aguilar NL = 1")
```
